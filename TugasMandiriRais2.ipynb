```python
# 1. Data diri Giant
import math

nama = "Giant"
usia = 20
berat = 97.5 #kg
tinggi = 1.75 #meter
status_olahraga = False
```

```python
# 2. Konversi String ke Float
berat_string = "97.5"
berat_float = float(berat_string)

# Kalimat pengantar
print("Halo, saya", nama + ".", "Usia saya", usia, "tahun dan berat saya", berat_float, "kg.")
```

```python
# 3. BMI Giant
BMI = berat_float / (tinggi**2)

print("BMI asli :", BMI)
print("BMI dibulatkan :", round(BMI, 2))
```

```python
# 4. Terget minum sehari
target_minum = 2000
botol = 350

hasil_asli = target_minum / botol
botol_penuh = target_minum // botol
sisa_air = target_minum % botol

print("Jumlah botol (hasil asli):", hasil_asli)
print("Jumlah botol penuh:", botol_penuh)
print("Sisa air(ml):", sisa_air, "ml")
```

```python
# 5. Kebutuhan Kalori

kalori = 2100.5
print("Kalori dibulatkan kebawah:", math.floor(kalori))
print("Kalori dibulatkan keatas:", math.ceil(kalori))
```
