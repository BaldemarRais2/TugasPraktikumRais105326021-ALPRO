```python
# 3. Nobita sering dimarahi oleh ibunya karena malas belajar dan sering menyontek. 
    # Doraemon akhirnya membuatkan “Program Peningkat Disiplin Harian”,
    # yaitu alat yang bisa menghitung skor kedisiplinan Nobita setiap hari berdasarkan aktivitasnya. 
    # Kamu akan membantu Doraemon membuat simulasi program tersebut menggunakan Python dan konsep fungsi.  
```

```python
#a. Membuat Variabel Global

skor_nobita = 0
```

```python
#b. Membuat Fungsi untuk Memperbarui Skor 

def update_skor(nilai):
    global skor_nobita
    skor_nobita += nilai
```

```python
#c. Mebuat fungsi Aktivitas Nobita
#1
def belajar(jam_belajar):
    if jam_belajar < 1:
        print("Nobita hanya membuka buku sebentar")
        return 10
    elif 1 <= jam_belajar < 3:
        print("Nobita belajar secukupnya")
        return 25
    else:
        print("Nobita belajar dengan rajin hari ini ")
        return 50

#2
def tidur(jam_tidur):
    if jam_tidur < 1:
        print("Nobita kurang tidur")
        return -10
    elif 1 <= jam_tidur <= 2:
        print("Waktu tidur Nobita seimbang")
        return 20
    else:
        print("Nobita tidur terlalu lama ")
        return -20

#3
def main_bersama_doraemon(jumlah_gadget, waktu_main):
    if waktu_main <= 1 and jumlah_gadget <= 2:
        print("Nobita bermain secukupnya")
        return 15
    elif waktu_main <= 2 and jumlah_gadget <= 3:
        print("Nobita masih bisa mengatur waktu dengan baik")
        return 5
    else:
        print("Nobita kebanyakan bermain ")
        return -30

#4
def menyontek():
    print("Nobita ketahuan menyontek")
    return -50
```

```python
#d. Membuat Fungsi Pengendali Aktivitas

def jalankan_aktivitas(pilihan):
    if pilihan == 1:
        jam_belajar = float(input("Berapa jam Nobita belajar hari ini?"))
        print(f"Jam belajar: {jam_belajar} jam")
        skor = belajar(jam_belajar)
        update_skor(skor)

    elif pilihan == 2:
        jam_tidur = float(input("Berapa jam Nobita tidur siang hari ini?"))
        print(f"Jam tidur: {jam_tidur} jam")
        skor = tidur(jam_tidur)
        update_skor(skor)

    elif pilihan == 3:
        jumlah_gadget = int(input("Berapa gadget yang digunakan Nobita?"))
        print("Jumlah Gadget:", jumlah_gadget)
        waktu_main = float(input("Berapa jam Nobita bermian gadget?"))
        print(f"Waktu main: {waktu_main} jam")
        skor = main_bersama_doraemon(jumlah_gadget, waktu_main)
        update_skor(skor)

    elif pilihan == 4:
        skor = menyontek()
        update_skor(skor)

    else :
        print("Pilihan tidak valid, aktivitas hanya ada 4 pilihan")
```

```python
#e. Membuat Fungsi Evaluasi Akhir
def hasil_evaluasi(skor_nobita):
    if skor_nobita >= 80:
        return "Siswa Teladan"
    elif 50 <= skor_nobita < 80:
        return "Cukup Rajin"
    else:
        return "Pemalas"
```

```python
#f. Membuat Menu Utama Program Peningkatan Disiplin Harian
print("=== PROGRAM PENINGKATAN DISIPILIN NOBITA ===")
print("1. Belajar")
print("2. Tidur")
print("3. Main bersama Doraemon")
print("4. Menyontek saat ujian")

print("")

pilihan = int(input("PIlih aktivitas (1-4):"))
jalankan_aktivitas(pilihan)
hasil = hasil_evaluasi(skor_nobita)
print(f"\nSkor Nobita: {skor_nobita}")
print(f"Evaluasi: {hasil}")
```
