```python
# Variable Parameter
usia_pengemudi = int(input("Usia: "))
jenis_kendaraan = input("Jenis kendaraan (mobil/motor): ").lower()
nilai_kendaraan = float(input("Masukkan nilai/harga kendaraan (Rp): "))
area_penggunaan = input("Lokasi penggunaan kendaraan (perkotaan/pedesaan): ").lower()
sistem_pengaman = input("Apakah kendaraan memiliki sistem pengaman? (ya/tidak): ").lower()

print(f"Usia: {usia_pengemudi} \n"
      f"Jenis Kendaraan: {jenis_kendaraan} \n"
      f"Nilai Kendaraan: {nilai_kendaraan} \n"
      f"Area penggunaan: {area_penggunaan} \n"
      f"Sistem_pengaman: {sistem_pengaman}")
```

```python
# Kategori Risiko (rendah, sedang, tinggi)
# Usia
if usia_pengemudi < 20 or usia_pengemudi > 60 :
    risiko = "Tinggi"
else:
    if jenis_kendaraan == "motor": # Jenis Kendaraan
        risiko = "Sedang"
    else :
        risiko = "Rendah"

print(f"Risiko: {risiko}")
```

```python
# Langkah Perhitungan Premi saya letakkan di bawah
# Area
if area_penggunaan == "perkotaan" :
    if risiko == "Rendah":
        risiko = "Sedang"
    elif risiko == "Sedang":
        risiko = "Tinggi"
# Else tidak saya tulis, karena tidak diberi instruksi apakah risiko akan berubah jika bernilai else

print(f"Risiko: {risiko}")
```

```python
# Sistem pengaman
if sistem_pengaman == "tidak":
    if risiko == "Rendah":
        risiko = "Sedang"
    elif risiko == "Sedang":
        risiko = "Tinggi"

print(f"Risiko: {risiko}")
```

```python
# Nilai & Premi
if risiko == "Tinggi":
    persentase = 0.1
elif risiko == "Sedang":
    persentase = 0.07
else :
    persentase = 0.5

premi = persentase * nilai_kendaraan
print(f"Premi: Rp.{premi:,.0f}")
```

```python
# Pesan (Conditional Expression)
pesan = ("Doraemon menyarankan asuransi tambahan untuk perlindungan penuh" if risiko == "Tinggi" else
         "Risiko kendaraan masih aman menurut doaremon")

# contoh output
print(f"Kategori Risiko: {risiko}")
print(f"Jumlah Premi yang perlu dibayar: Rp.{premi:,.0f}")
print(f"Pesan Doraemon: {pesan}")
```
