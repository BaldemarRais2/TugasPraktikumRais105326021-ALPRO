```python
hari = 1
saldo = 50000

while hari <= 30:
    pengeluaran = int(input("Pengeluaran:"))
    print(f"Pengeluaran hari ke-{hari}: {pengeluaran}")
    saldo -= pengeluaran
    if saldo < 0:
        break
    print(f"Sisa saldo hari ke-{hari}: {saldo}")
    hari += 1
if saldo < 0:
    print(f"Saldo habis di hari ke-{hari}")
```

```python
for bulan in range(1, 4):
    print(f"\nBulan ke-{bulan}")
    total_sebulan = 0

    for minggu in range(1, 5):
        pengeluaran_minggu = int(input("Pengeluaran:"))
        print(f"  Pengeluaran minggu ke-{minggu}, {pengeluaran_minggu}")

        if pengeluaran_minggu > 20000:
            print("Pengeluaran terlalu besar! Pencatatan bulan ini dihentikan")
            break

        total_sebulan += pengeluaran_minggu
    else:
        print(f"Total pengeluaran bulan ke-{bulan}: Rp{total_sebulan}")
```

```python
saldo_awal = 300000
saldo = saldo_awal

for bulan in range(1, 13):
    print(f"Bulan ke-{bulan}")
    minggu = 1
    while minggu <= 4:
        libur = input("Apakah minggu ini libur?(y/n):")

        if libur.lower() == "y":
            minggu += 1
            continue
        else:
            pengeluaran = int(input("pengeluaran:"))
            saldo -= pengeluaran

        if saldo < 0 :
            print("saldo kurang dari 0")
            break
        minggu += 1
        
    else:
        print(f"Saldo sisa akhir bulan ke-{bulan} sebesar Rp{saldo}")
        continue
    break
else:
    print(f"Nobita berhasil menabung Rp{saldo} dari total Rp{saldo_awal}")
```
