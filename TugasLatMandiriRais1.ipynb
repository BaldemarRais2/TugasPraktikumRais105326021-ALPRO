# Praktikum: Toko Dorayaki Doraemon
    # Deskripsi singkat: Program ini dibuat untuk membantu Doraemon dan teman-temannya membuka usaha Toko Dorayaki di Tokyo. Program ini meminta input berupa nama makanan, karakter, dan sifat untuk membentuk nama toko. Selain itu, program menghitung total modal, pendapatan dari penjualan, serta keuntungan. Program juga menyertakan contoh error yang dibuat Nobita, sehingga dapat menjadi bahan pembelajaran bagi pemula seperti Shizuka dalam memahami dasar-dasar Python.

```python
# 1. Meminta input dari user untuk nama toko
makanan = input("Nama makanan: ")   # Variabel nama makanan
karakter = input("Nama karakter: ") # Variabel nama karakter
sifat = input("Kata sifat: ")       # Variabel kata sifat
```

```python
# 2. Menggabungkan input menjadi nama toko
print(makanan, karakter, sifat, sep=" ")
```

```python
# 3. Modal toko dari setiap karakter
doraemon = 555000   # modal doraemon
nobita = 333000     # modal nobita
shizuka = 222000    # modal shizuka
```

```python
# 4. Hitung todal modal terkumpul
total_modal = doraemon + nobita + shizuka
print("Total modal terkumpul: Rp", total_modal)
```

```python
# 5. Hitung pendapatan dari penjualan Dorayaki
jumlah_dorayaki = 350   #  variabel jumlah dorayaki
harga_dorayaki = 5000   # variabel harga dorayaki
total_pendapatan = jumlah_dorayaki * harga_dorayaki
print("Total pendapatan: Rp", total_pendapatan)
```

```python
#6. Hitung keuntungan penjualan Dorayaki bulan pertama
keuntungan = total_pendapatan - total_modal
print("Keuntungan Dorayaki bulan pertama: Rp", keuntungan)
```

```python
#7. Contoh penulisan Nobita yang salah
# Nobita menulis:
print("Keuntungan total adalah" keuntungan)
```
