```python
# 1. Membuat data belanjaan
# List belanjaan
belanjaan = ["Apel", "Susu", "Roti"]
# Tuple harga & jumlah
harga_jumlah = (5000, 3, 12000, 2, 8000, 1)
# Print karakter
print(belanjaan[0][0])
print(belanjaan[-1][-1])
```

```python
# 2. Hitung Total Harga per Barang
# Total harga masing masing
harga_apel = harga_jumlah[0] * harga_jumlah[1]
harga_susu = harga_jumlah[2] * harga_jumlah[3]
harga_roti = harga_jumlah[4] * harga_jumlah[5]
harga_total = harga_apel + harga_susu + harga_roti
# Total harga keseluruhan
print(f"Harga total Apel: Rp.{harga_apel}")
print(f"Harga total Susu: Rp.{harga_susu}")
print(f"Harga total Roti: Rp.{harga_roti}")
print(" ")
print(f"Harga Total Keseluruhan: Rp.{harga_total}")
```
